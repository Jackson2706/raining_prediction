```python
import joblib
import os
```

```python
weight_path = "/home/jackson-devworks/Desktop/raining_prediction/weights/ThacXang_xgb.pkl"
model = joblib.load(weight_path)
model_name = os.path.basename(weight_path).replace(".pkl", "")
model_list = ["COMS","GFS","MITSUISHI2011_D02","LING3_D02","LINKF_D02","LINBMJ_D02","ETAKF_D02","ETAG3_D02","ETABMJ_D02"]
features = ["Time_difference/Time Interval"] + model_list
```

# 1. Feature Importance

```python
import matplotlib.pyplot as plt
```

```python
importances = model.feature_importances_
plt.barh(features, importances)
plt.xlabel('Feature Importance')
plt.ylabel('Feature')
plt.title(f'Feature Importance in {model_name}')
plt.savefig(os.path.join("/home/jackson-devworks/Desktop/raining_prediction/figures", "feature_" + os.path.basename(weight_path).replace(".pkl", ".png")), bbox_inches='tight',  # Cắt sát nội dung của hình
    pad_inches=0.5  # Thêm padding xung quanh hình)
)
plt.show()
```

# 2. Tree Visualization

```python
from sklearn.tree import plot_tree
```

```python
plt.figure(figsize=(20, 10))
plot_tree(model.estimators_[0], feature_names=features, filled=True, max_depth=3)
plt.savefig(
    os.path.join("/home/jackson-devworks/Desktop/raining_prediction/figures", 
    "tree_visualization_" + os.path.basename(weight_path).replace(".pkl", ".png")), 
    bbox_inches='tight',  # Cắt sát nội dung của hình
    pad_inches=0.5,  # Thêm padding xung quanh hình
    dpi=300  # Tăng độ phân giải lên 300 DPI (hoặc cao hơn)
)
plt.show()
```
